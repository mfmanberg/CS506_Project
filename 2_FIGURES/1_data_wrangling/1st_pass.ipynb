```python
import os
import csv
from datetime import datetime
import pandas as pd

def get_date_from_csv(csv_path):
    """Extract the first and last dates from a CSV file."""
    try:
        # Read just the first few rows to get the first date
        with open(csv_path, 'r', encoding='utf-8') as file:
            reader = csv.reader(file)
            header = next(reader)  # Skip header
            first_row = next(reader)
            first_timestamp = first_row[0]
        
        # Read the last few rows to get the last date
        # For efficiency, we'll assume single-day files and just use the first date
        # Parse the timestamp format "MM/DD/YYYY HH:MM:SS"
        date_part = first_timestamp.split(' ')[0]  # Get just the date part
        return date_part, date_part  # Return same date for both first and last
        
    except Exception as e:
        print(f"Error reading {csv_path}: {e}")
        return None, None

def rename_csv_files(directory):
    """Rename all CSV files in the directory based on their internal dates."""
    
    # Get all CSV files in the directory
    csv_files = [f for f in os.listdir(directory) if f.endswith('.csv')]
    
    print(f"Found {len(csv_files)} CSV files to process...")
    
    renamed_count = 0
    errors = []
    
    for csv_file in csv_files:
        csv_path = os.path.join(directory, csv_file)
        
        # Get the dates from the CSV
        first_date, last_date = get_date_from_csv(csv_path)
        
        if first_date and last_date:
            # Since files contain single-day data, first_date equals last_date
            # Format: MM/DD/YYYY -> MM_DD_YYYY for filename (replace / with _)
            formatted_date = first_date.replace('/', '_')
            new_filename = f"{formatted_date}.csv"
            new_path = os.path.join(directory, new_filename)
            
            # Check if file already exists with the new name
            if os.path.exists(new_path) and new_path != csv_path:
                print(f"Warning: {new_filename} already exists, skipping {csv_file}")
                continue
            
            # Rename the file
            try:
                if csv_path != new_path:  # Only rename if different
                    os.rename(csv_path, new_path)
                    print(f"Renamed: {csv_file} -> {new_filename}")
                    renamed_count += 1
                else:
                    print(f"Already correctly named: {csv_file}")
            except Exception as e:
                error_msg = f"Error renaming {csv_file}: {e}"
                print(error_msg)
                errors.append(error_msg)
        else:
            error_msg = f"Could not extract dates from {csv_file}"
            print(error_msg)
            errors.append(error_msg)
    
    print(f"\nRenaming complete!")
    print(f"Successfully renamed: {renamed_count} files")
    if errors:
        print(f"Errors encountered: {len(errors)}")
        for error in errors[:5]:  # Show first 5 errors
            print(f"  - {error}")
        if len(errors) > 5:
            print(f"  ... and {len(errors) - 5} more errors")

if __name__ == "__main__":
    # Directory containing the CSV files
    csv_directory = r"c:\Users\Matt\Desktop\CS506\CS506_Project\nyiso_csvs"
    
    # Confirm directory exists
    if not os.path.exists(csv_directory):
        print(f"Error: Directory {csv_directory} does not exist!")
    else:
        print(f"Processing CSV files in: {csv_directory}")
        rename_csv_files(csv_directory)
```

### SORT CSVs BY YEAR

```python
import os
import shutil
from datetime import datetime
import pandas as pd


def sort_csvs_by_year():
    """Sort CSV files from nyiso_csvs into yearly folders in nyiso_yearly."""
    
    # Define source and destination directories
    source_dir = r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_csv"
    dest_dir = r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_csv"
    
    # Create destination directory if it doesn't exist
    os.makedirs(dest_dir, exist_ok=True)
    
    # Check if source directory exists
    if not os.path.exists(source_dir):
        print(f"Error: Source directory {source_dir} does not exist!")
        return
    
    # Get all CSV files
    csv_files = [f for f in os.listdir(source_dir) if f.endswith('.csv')]
    print(f"Found {len(csv_files)} CSV files to sort...")
    
    # Dictionary to track files by year
    files_by_year = {}
    errors = []
    
    for csv_file in csv_files:
        try:
            # Extract year from filename (format: MM_DD_YYYY.csv)
            filename_without_ext = csv_file.replace('.csv', '')
            date_parts = filename_without_ext.split('_')
            
            if len(date_parts) == 3:
                month, day, year = date_parts
                year = int(year)
                
                # Add to tracking dictionary
                if year not in files_by_year:
                    files_by_year[year] = []
                files_by_year[year].append(csv_file)
                
            else:
                errors.append(f"Could not parse date from filename: {csv_file}")
                
        except Exception as e:
            errors.append(f"Error processing {csv_file}: {e}")
    
    # Sort years and process files
    sorted_years = sorted(files_by_year.keys())
    total_moved = 0
    
    for year in sorted_years:
        year_dir = os.path.join(dest_dir, str(year))
        os.makedirs(year_dir, exist_ok=True)
        
        # Sort files within the year by date
        year_files = files_by_year[year]
        year_files.sort(key=lambda x: datetime.strptime(x.replace('.csv', '').replace('_', '/'), '%m/%d/%Y'))
        
        print(f"\nProcessing year {year} ({len(year_files)} files):")
        
        for csv_file in year_files:
            source_path = os.path.join(source_dir, csv_file)
            dest_path = os.path.join(year_dir, csv_file)
            
            try:
                # Copy file to year folder
                shutil.copy2(source_path, dest_path)
                print(f"  Copied: {csv_file}")
                total_moved += 1
                
            except Exception as e:
                error_msg = f"Error copying {csv_file}: {e}"
                print(f"  {error_msg}")
                errors.append(error_msg)
    
    # Summary
    print(f"\n{'='*50}")
    print(f"Sorting complete!")
    print(f"Years processed: {len(sorted_years)} ({min(sorted_years) if sorted_years else 'N/A'} - {max(sorted_years) if sorted_years else 'N/A'})")
    print(f"Total files copied: {total_moved}")
    
    if errors:
        print(f"Errors encountered: {len(errors)}")
        for error in errors[:5]:  # Show first 5 errors
            print(f"  - {error}")
        if len(errors) > 5:
            print(f"  ... and {len(errors) - 5} more errors")
    
    # Show folder structure
    print(f"\nCreated folder structure:")
    for year in sorted_years:
        year_dir = os.path.join(dest_dir, str(year))
        file_count = len([f for f in os.listdir(year_dir) if f.endswith('.csv')])
        print(f"  {year}/: {file_count} files")

# Run the sorting function
sort_csvs_by_year()

print(f"\nDone:")
```

### AGGREGATE NYISO_YEARLY 
Make a new subfolder of LIB called nyiso_yearly, combine each year's files into one csvs per year 

```python
import os
import pandas as pd
from datetime import datetime

def combine_yearly_csvs():
    """Create LIB/nyiso_yearly folder and combine each year's CSV files into one file per year"""
    
    # Define source and destination directories
    source_dir = r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_csv"
    dest_dir = r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_csv\nyiso_yearly"
    
    # Create destination directory if it doesn't exist
    os.makedirs(dest_dir, exist_ok=True)
    
    # Check if source directory exists
    if not os.path.exists(source_dir):
        print(f"Error: Source directory {source_dir} does not exist!")
        return
    
    # Get all year folders in the source directory
    year_folders = []
    for item in os.listdir(source_dir):
        item_path = os.path.join(source_dir, item)
        if os.path.isdir(item_path) and item.isdigit() and len(item) == 4:
            year_folders.append(item)
    
    year_folders.sort()
    print(f"Found {len(year_folders)} year folders: {year_folders}")
    
    if not year_folders:
        print("No year folders found to process!")
        return
    
    # Process each year
    for year in year_folders:
        year_path = os.path.join(source_dir, year)
        csv_files = [f for f in os.listdir(year_path) if f.endswith('.csv')]
        
        if not csv_files:
            print(f"No CSV files found in {year} folder")
            continue
        
        print(f"\nProcessing year {year} with {len(csv_files)} files...")
        
        # Sort files by date for proper chronological order
        csv_files.sort(key=lambda x: datetime.strptime(x.replace('.csv', '').replace('_', '/'), '%m/%d/%Y'))
        
        # List to store all dataframes for this year
        yearly_dataframes = []
        
        # Read each CSV file
        for csv_file in csv_files:
            file_path = os.path.join(year_path, csv_file)
            try:
                df = pd.read_csv(file_path)
                yearly_dataframes.append(df)
                print(f"  Read: {csv_file} ({len(df)} rows)")
            except Exception as e:
                print(f"  Error reading {csv_file}: {e}")
                continue
        
        # Combine all dataframes for this year
        if yearly_dataframes:
            try:
                combined_df = pd.concat(yearly_dataframes, ignore_index=True)
                
                # Sort by timestamp if timestamp column exists
                if len(combined_df.columns) > 0:
                    timestamp_col = combined_df.columns[0]  # Assume first column is timestamp
                    try:
                        combined_df[timestamp_col] = pd.to_datetime(combined_df[timestamp_col])
                        combined_df = combined_df.sort_values(timestamp_col)
                        combined_df[timestamp_col] = combined_df[timestamp_col].astype(str)
                    except:
                        pass  # Keep original format if datetime conversion fails
                
                # Save combined file
                output_filename = f"{year}_combined.csv"
                output_path = os.path.join(dest_dir, output_filename)
                combined_df.to_csv(output_path, index=False)
                
                print(f"  ✅ Created: {output_filename} ({len(combined_df)} total rows)")
                
            except Exception as e:
                print(f"  Error combining files for {year}: {e}")
        else:
            print(f"  No valid data found for {year}")
    
    # Summary
    print(f"\n{'='*60}")
    print(f"Yearly combination complete!")
    print(f"Output directory: {dest_dir}")
    
    # Show final structure
    if os.path.exists(dest_dir):
        output_files = [f for f in os.listdir(dest_dir) if f.endswith('.csv')]
        output_files.sort()
        print(f"Created {len(output_files)} yearly files:")
        
        for file in output_files:
            file_path = os.path.join(dest_dir, file)
            try:
                df = pd.read_csv(file_path)
                print(f"  {file}: {len(df):,} rows")
            except:
                print(f"  {file}: Error reading file")

# Run the function
combine_yearly_csvs()
```

```python
import os
from pathlib import Path
import shutil

def gather_top_level_csvs(
    source_dir=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_csv",
    dest_dir=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_csv\nyiso_all",
    move: bool = True,
    dry_run: bool = False,
    overwrite: bool = True,
    verbose: bool = True
):
    """
    Move or copy CSV files that are directly inside `source_dir` (not inside any subfolders)
    into `dest_dir` as individual files (no aggregation).

    Parameters:
      - source_dir: folder containing nyiso CSVs (top-level files considered)
      - dest_dir: folder where top-level CSVs will be copied/moved
      - move: if True, move files; if False, copy files
      - dry_run: if True, only print planned actions and return summary (no fs changes)
      - overwrite: if True, overwrite existing files in dest_dir
      - verbose: if True, print progress

    Returns:
      summary dict: {files_found, files_transferred, skipped, errors}
    """
    src = Path(source_dir).expanduser().resolve()
    dst = Path(dest_dir).expanduser().resolve()
    summary = {"files_found": 0, "files_transferred": 0, "skipped": 0, "errors": []}

    if not src.exists():
        msg = f"Source directory not found: {src}"
        summary["errors"].append(msg)
        if verbose:
            print(msg)
        return summary

    # create destination (but don't create it when dry_run is True unless necessary to show path)
    if not dry_run:
        dst.mkdir(parents=True, exist_ok=True)

    # collect only top-level csv files (do not recurse)
    top_level = [p for p in src.iterdir() if p.is_file() and p.suffix.lower() == ".csv"]
    summary["files_found"] = len(top_level)
    if verbose:
        print(f"Found {len(top_level)} top-level CSV file(s) in {src}")

    if dry_run:
        if verbose:
            print("DRY RUN - actions shown below (no files will be moved/copied):")
            for p in top_level:
                print(f"  - {p.name} -> {dst / p.name}")
        return summary

    for p in top_level:
        dest_path = dst / p.name
        try:
            if dest_path.exists():
                if overwrite:
                    # remove existing before action
                    dest_path.unlink()
                else:
                    summary["skipped"] += 1
                    if verbose:
                        print(f"Skipping existing file: {dest_path.name}")
                    continue

            if move:
                shutil.move(str(p), str(dest_path))
                action = "Moved"
            else:
                shutil.copy2(str(p), str(dest_path))
                action = "Copied"

            summary["files_transferred"] += 1
            if verbose:
                print(f"{action}: {p.name} -> {dest_path}")
        except Exception as e:
            msg = f"Error handling {p.name}: {e}"
            summary["errors"].append(msg)
            if verbose:
                print(msg)

    if verbose:
        print(f"Done. {summary['files_transferred']} files {'moved' if move else 'copied'}; {summary['skipped']} skipped; {len(summary['errors'])} errors.")
    return summary
```

```python
import os
from pathlib import Path
import pandas as pd

def combine_all_csvs_to_master(
    source_dir=r"C:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_csv\nyiso_all",
    dest_dir=r"C:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_csv\nyiso_master",
    output_name="nyiso_master.csv",
    dry_run=False,            # default safe mode
    overwrite=False,
    chunksize=100000,
    verbose=True,
):
    # initialize paths and summary
    src = Path(source_dir).expanduser().resolve()
    dst = Path(dest_dir).expanduser().resolve()
    summary = {"files_found": 0, "files_combined": 0, "skipped": 0, "errors": []}

    if not src.exists():
        msg = f"Source directory not found: {src}"
        summary["errors"].append(msg)
        if verbose:
            print(msg)
        return summary

    # collect only top-level csv files (do not recurse)
    csv_files = sorted([p for p in src.iterdir() if p.is_file() and p.suffix.lower() == ".csv"])
    summary["files_found"] = len(csv_files)

    if verbose:
        print(f"Found {len(csv_files)} CSV file(s) in {src}")

    if len(csv_files) == 0:
        return summary

    out_path = dst / output_name

    if dry_run:
        if verbose:
            print("DRY RUN - planned actions:")
            print(f"  - destination: {out_path}")
            if out_path.exists():
                print(f"  - NOTE: output already exists: {out_path} (overwrite={overwrite})")
            print("  - files to be combined (showing up to 50):")
            for p in csv_files[:50]:
                print(f"    - {p.name}")
            if len(csv_files) > 50:
                print(f"    ... and {len(csv_files)-50} more files")
        return summary

    # ensure destination exists
    dst.mkdir(parents=True, exist_ok=True)

    # Check for existing output file
    if out_path.exists():
        if not overwrite:
            msg = f"Output file already exists: {out_path} (set overwrite=True to replace)"
            summary["errors"].append(msg)
            if verbose:
                print(msg)
            return summary
        else:
            try:
                out_path.unlink()
                if verbose:
                    print(f"Removed existing output: {out_path}")
            except Exception as e:
                msg = f"Unable to remove existing output: {e}"
                summary["errors"].append(msg)
                if verbose:
                    print(msg)
                return summary

    # Build master column list by scanning headers only (fast, low memory)
    master_cols = []
    for p in csv_files:
        try:
            cols = list(pd.read_csv(p, nrows=0).columns)
            if not master_cols:
                master_cols = cols.copy()
            else:
                for c in cols:
                    if c not in master_cols:
                        master_cols.append(c)
        except Exception as e:
            msg = f"Error reading header from {p.name}: {e}"
            summary["errors"].append(msg)
            if verbose:
                print(msg)

    if verbose:
        print(f"Master column count: {len(master_cols)}")

    # Stream files and append to output with consistent columns
    first_write = True
    combined_count = 0

    for p in csv_files:
        try:
            if chunksize:
                for chunk in pd.read_csv(p, chunksize=chunksize):
                    chunk = chunk.reindex(columns=master_cols)
                    if first_write:
                        chunk.to_csv(out_path, index=False, mode="w", header=True)
                        first_write = False
                    else:
                        chunk.to_csv(out_path, index=False, mode="a", header=False)
                    combined_count += len(chunk)
            else:
                df = pd.read_csv(p)
                df = df.reindex(columns=master_cols)
                if first_write:
                    df.to_csv(out_path, index=False, mode="w", header=True)
                    first_write = False
                else:
                    df.to_csv(out_path, index=False, mode="a", header=False)
                combined_count += len(df)
            if verbose:
                print(f"Appended: {p.name} ({p.stat().st_size:,} bytes)")
        except Exception as e:
            msg = f"Error combining {p.name}: {e}"
            summary["errors"].append(msg)
            if verbose:
                print(msg)

    summary["files_combined"] = len(csv_files) if combined_count >= 0 else 0
    if verbose:
        print(f"Done. Wrote {combined_count:,} rows to {out_path} ({len(summary['errors'])} errors)")
    return summary

# Run the real combine (writes master CSV)
res = combine_all_csvs_to_master(dry_run=False, overwrite=False, verbose=True)
import json
print("Combine result:\n", json.dumps(res, indent=2))
```

```python
# Convert all CSVs under nyiso_csv -> nyiso_parquet (dry-run by default)
import os
from pathlib import Path
import pandas as pd


def convert_csvs_to_parquet(
    source_root=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_csv",
    dest_root=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_parquet",
    dry_run= False,
    overwrite=False,
    verbose=True,
    max_files=None,  # limit for testing
    preview_limit=50,  # number of file examples to show on dry-run
):
    """Recursively copy folder structure from source_root into dest_root and convert each .csv to .parquet.

    - dry_run=True will only print planned actions and return a summary.
    - overwrite controls whether existing .parquet files are replaced.
    - max_files can be set for testing to limit actual work.
    """
    src = Path(source_root).expanduser().resolve()
    dst = Path(dest_root).expanduser().resolve()
    summary = {"files_found": 0, "converted": 0, "skipped": 0, "errors": []}

    if not src.exists():
        msg = f"Source directory not found: {src}"
        summary["errors"].append(msg)
        if verbose:
            print(msg)
        return summary

    csv_paths = sorted(src.rglob("*.csv"))
    summary["files_found"] = len(csv_paths)

    if verbose:
        print(f"Found {len(csv_paths)} CSV file(s) under {src}")

    if len(csv_paths) == 0:
        return summary

    # show preview when dry-run
    if dry_run:
        print("DRY RUN - planned conversion (no files changed):")
        print(f"  - source root: {src}")
        print(f"  - dest root:   {dst}")
        if len(csv_paths) > 0:
            print(f"  - showing up to {preview_limit} files:")
            for p in csv_paths[:preview_limit]:
                rel = p.relative_to(src)
                out_path = (dst / rel).with_suffix('.parquet')
                print(f"    - {p}  ->  {out_path}")
            if len(csv_paths) > preview_limit:
                print(f"    ... and {len(csv_paths)-preview_limit} more files")
        return summary

    # ensure destination exists (we'll create subdirectories as needed)
    dst.mkdir(parents=True, exist_ok=True)

    processed = 0
    for p in csv_paths:
        if max_files is not None and processed >= max_files:
            break
        try:
            rel = p.relative_to(src)
            out_dir = (dst / rel).parent
            out_dir.mkdir(parents=True, exist_ok=True)
            out_file = (dst / rel).with_suffix('.parquet')

            if out_file.exists():
                if not overwrite:
                    summary["skipped"] += 1
                    if verbose:
                        print(f"Skipping existing: {out_file}")
                    processed += 1
                    continue
                else:
                    try:
                        out_file.unlink()
                    except Exception as e:
                        msg = f"Unable to remove existing output {out_file}: {e}"
                        summary["errors"].append(msg)
                        if verbose:
                            print(msg)
                        processed += 1
                        continue

            # read csv and write parquet
            try:
                df = pd.read_csv(p)
            except Exception as e:
                msg = f"Error reading CSV {p}: {e}"
                summary["errors"].append(msg)
                if verbose:
                    print(msg)
                processed += 1
                continue

            try:
                # Try default engine (pyarrow preferred)
                df.to_parquet(out_file, index=False)
            except Exception as e1:
                # attempt fallback to fastparquet if available
                try:
                    df.to_parquet(out_file, index=False, engine="fastparquet")
                except Exception as e2:
                    msg = f"Error writing parquet for {p}: {e1} | {e2}"
                    summary["errors"].append(msg)
                    if verbose:
                        print(msg)
                    processed += 1
                    continue

            summary["converted"] += 1
            if verbose:
                print(f"Converted: {p} -> {out_file}")

        except Exception as e:
            msg = f"Unexpected error processing {p}: {e}"
            summary["errors"].append(msg)
            if verbose:
                print(msg)
        processed += 1

    if verbose:
        print(f"Done. Converted: {summary['converted']}; Skipped: {summary['skipped']}; Errors: {len(summary['errors'])}")
    return summary


# Run a dry-run preview so you can inspect planned actions before converting.
_preview = convert_csvs_to_parquet(dry_run=False, preview_limit=50, verbose=True)
import json
print('\nPreview summary:\n', json.dumps(_preview, indent=2))



```

```python
import os
from pathlib import Path
import shutil

def gather_top_level_csvs(
    source_dir = r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\mesonet",
    dest_dir = r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\mesonet\mesonet_csv",
    move: bool = True,
    dry_run: bool = False,
    overwrite: bool = False,
    verbose: bool = True
):
    """
    Move or copy CSV files that are directly inside `source_dir` (not inside any subfolders)
    into `dest_dir` as individual files (no aggregation).

    Parameters:
      - source_dir: folder containing nyiso CSVs (top-level files considered)
      - dest_dir: folder where top-level CSVs will be copied/moved
      - move: if True, move files; if False, copy files
      - dry_run: if True, only print planned actions and return summary (no fs changes)
      - overwrite: if True, overwrite existing files in dest_dir
      - verbose: if True, print progress

    Returns:
      summary dict: {files_found, files_transferred, skipped, errors}
    """
    src = Path(source_dir).expanduser().resolve()
    dst = Path(dest_dir).expanduser().resolve()
    summary = {"files_found": 0, "files_transferred": 0, "skipped": 0, "errors": []}

    if not src.exists():
        msg = f"Source directory not found: {src}"
        summary["errors"].append(msg)
        if verbose:
            print(msg)
        return summary

    # create destination (but don't create it when dry_run is True unless necessary to show path)
    if not dry_run:
        dst.mkdir(parents=True, exist_ok=True)

    # collect only top-level csv files (do not recurse)
    top_level = [p for p in src.iterdir() if p.is_file() and p.suffix.lower() == ".csv"]
    summary["files_found"] = len(top_level)
    if verbose:
        print(f"Found {len(top_level)} top-level CSV file(s) in {src}")

    if dry_run:
        if verbose:
            print("DRY RUN - actions shown below (no files will be moved/copied):")
            for p in top_level:
                print(f"  - {p.name} -> {dst / p.name}")
        return summary

    for p in top_level:
        dest_path = dst / p.name
        try:
            if dest_path.exists():
                if overwrite:
                    # remove existing before action
                    dest_path.unlink()
                else:
                    summary["skipped"] += 1
                    if verbose:
                        print(f"Skipping existing file: {dest_path.name}")
                    continue

            if move:
                shutil.move(str(p), str(dest_path))
                action = "Moved"
            else:
                shutil.copy2(str(p), str(dest_path))
                action = "Copied"

            summary["files_transferred"] += 1
            if verbose:
                print(f"{action}: {p.name} -> {dest_path}")
        except Exception as e:
            msg = f"Error handling {p.name}: {e}"
            summary["errors"].append(msg)
            if verbose:
                print(msg)

    if verbose:
        print(f"Done. {summary['files_transferred']} files {'moved' if move else 'copied'}; {summary['skipped']} skipped; {len(summary['errors'])} errors.")
    return summary

result = gather_top_level_csvs(dry_run=False, move=True, overwrite=False, verbose=True)
#print(json.dumps(result, indent=2))
```

```python
import os
import csv
import pandas as pd
import re
from datetime import datetime, timedelta

# Directory containing mesonet CSV files (top-level folder)
csv_directory = r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\mesonet\mesonet_csv"

# Timezone offset mappings (hours from UTC)
TIMEZONE_OFFSETS = {
    'EDT': -4,  # Eastern Daylight Time
    'EST': -5,  # Eastern Standard Time
    'CDT': -5,  # Central Daylight Time
    'CST': -6,  # Central Standard Time
    'MDT': -6,  # Mountain Daylight Time
    'MST': -7,  # Mountain Standard Time
    'PDT': -7,  # Pacific Daylight Time
    'PST': -8,  # Pacific Standard Time
}


def is_properly_named(filename):
    """Check if filename is already in MM_DD_YYYY.csv format."""
    # Pattern: MM_DD_YYYY.csv (with optional leading zeros)
    pattern = r'^\d{2}_\d{2}_\d{4}\.csv$'
    return bool(re.match(pattern, filename, re.IGNORECASE))


def _parse_ts_to_utc(ts_str):
    """Try several common timestamp formats and return a UTC datetime object or None."""
    if not ts_str:
        return None
    s = ts_str.strip()
    
    # Check if there's a timezone abbreviation at the end
    timezone_abbrev = None
    for tz in TIMEZONE_OFFSETS.keys():
        if s.endswith(' ' + tz):
            timezone_abbrev = tz
            s_without_tz = s[:-len(tz)-1].strip()
            break
    else:
        s_without_tz = s
    
    # Common timestamp formats to try (without timezone)
    fmts = [
        '%Y-%m-%d %H:%M:%S',
        '%Y-%m-%d',
        '%Y%m%d',
        '%m/%d/%Y %H:%M:%S',
        '%m/%d/%Y %H:%M',
        '%m/%d/%Y',
        '%m-%d-%Y %H-%M-%S',
        '%m-%d-%Y',
    ]
    
    dt = None
    for f in fmts:
        try:
            dt = datetime.strptime(s_without_tz, f)
            break
        except Exception:
            pass
    
    # If parsing failed, try regex fallbacks
    if dt is None:
        
        # Try MM/DD/YYYY or M/D/YYYY format
        m = re.search(r"(?P<m>\d{1,2})[/-](?P<d>\d{1,2})[/-](?P<y>\d{4})", s_without_tz)
        if m:
            try:
                dt = datetime(int(m.group('y')), int(m.group('m')), int(m.group('d')))
            except Exception:
                pass
        
        # Try YYYYMMDD format
        if dt is None:
            m2 = re.search(r"(?P<y>(19|20)\d{2})(?P<m>\d{2})(?P<d>\d{2})", s_without_tz)
            if m2:
                try:
                    dt = datetime(int(m2.group('y')), int(m2.group('m')), int(m2.group('d')))
                except Exception:
                    pass
    
    if dt is None:
        return None
    
    # Convert to UTC if timezone was detected
    if timezone_abbrev:
        offset_hours = TIMEZONE_OFFSETS[timezone_abbrev]
        # Subtract the offset to get UTC (e.g., EDT is UTC-4, so add 4 hours to get UTC)
        dt_utc = dt - timedelta(hours=offset_hours)
        return dt_utc
    
    # If no timezone, assume it's already in the desired timezone
    return dt


def convert_timestamp_to_utc_string(timestamp_str):
    """Convert a timestamp string with timezone to UTC and return without timezone label."""
    if not timestamp_str or not isinstance(timestamp_str, str):
        return timestamp_str
    
    dt_utc = _parse_ts_to_utc(timestamp_str)
    if dt_utc:
        # Return as string without timezone label
        return dt_utc.strftime('%Y-%m-%d %H:%M:%S')
    
    # If parsing failed, return original
    return timestamp_str


def process_csv_remove_timezone(csv_path, inplace=True, rename_time_column=True):
    """Process a CSV file to convert all timestamps to UTC and remove timezone labels.
    
    Args:
        csv_path: Path to the CSV file
        inplace: If True, overwrites the original file. If False, returns the DataFrame.
        rename_time_column: If True, renames 'time' column to 'datetime'
    """
    try:
        # Read the CSV file
        df = pd.read_csv(csv_path, low_memory=False)
        
        # Find columns that likely contain timestamps
        time_columns = []
        for col in df.columns:
            if 'time' in col.lower() or 'date' in col.lower():
                time_columns.append(col)
        
        # Convert timestamps in identified columns
        for col in time_columns:
            df[col] = df[col].apply(convert_timestamp_to_utc_string)
        
        # Rename 'time' column to 'datetime' if requested
        if rename_time_column and 'time' in df.columns:
            df = df.rename(columns={'time': 'datetime'})
        
        if inplace:
            # Overwrite the original file
            df.to_csv(csv_path, index=False)
            return True, None
        else:
            return df, None
    except Exception as e:
        return None, str(e)


def get_date_from_csv(csv_path):
    """Extract the first and last dates from a CSV file (converted to UTC).

    Looks for common time-like header names (case-insensitive): 'time', 'datetime', 'timestamp', 'time stamp'.
    Falls back to the first column if none found. Returns (MM/DD/YYYY, MM/DD/YYYY) or (None, None).
    """
    try:
        first_dt = None
        last_dt = None
        with open(csv_path, 'r', encoding='utf-8', errors='replace', newline='') as fh:
            # Try DictReader to handle headers
            reader = csv.DictReader(fh)
            fieldnames = reader.fieldnames or []
            
            # detect time-like header
            time_col = None
            want = {'time', 'datetime', 'timestamp', 'time stamp'}
            for fn in fieldnames:
                if fn and fn.strip().lower() in want:
                    time_col = fn
                    break
            
            # if not found, also consider headers that contain 'time' or 'date'
            if time_col is None:
                for fn in fieldnames:
                    if fn and ('time' in fn.lower() or 'date' in fn.lower()):
                        time_col = fn
                        break
            
            # if still not found, fallback to first column index
            if not fieldnames:
                # no header scenario - reset and use simple reader
                fh.seek(0)
                rdr = csv.reader(fh)
                for row in rdr:
                    if row and len(row) > 0 and row[0].strip():
                        dt = _parse_ts_to_utc(row[0].strip())
                        if dt:
                            first_dt = dt
                            last_dt = dt
                            break
                # continue to end to find last
                for row in rdr:
                    if row and len(row) > 0 and row[0].strip():
                        dt = _parse_ts_to_utc(row[0].strip())
                        if dt:
                            last_dt = dt
            else:
                # iterate rows using DictReader
                for row in reader:
                    # determine value to parse
                    val = None
                    if time_col:
                        val = row.get(time_col)
                    else:
                        # fallback to first column value
                        first_field = fieldnames[0]
                        val = row.get(first_field)
                    if val and str(val).strip():
                        dt = _parse_ts_to_utc(str(val).strip())
                        if dt:
                            if first_dt is None:
                                first_dt = dt
                            last_dt = dt
        
        if first_dt is None:
            return None, None
        
        # Extract just the date portion for filename
        first_date = first_dt.date()
        last_date = last_dt.date() if last_dt else first_date
        
        return first_date.strftime('%m/%d/%Y'), last_date.strftime('%m/%d/%Y')
    except Exception as e:
        print(f"Error reading {csv_path}: {e}")
        return None, None


def rename_csv_files(directory, dry_run=True, verbose=True, remove_timezone=True, rename_time_column=True):
    """Rename all CSV files in the directory based on their internal dates (UTC-adjusted).
    
    Args:
        directory: Path to the directory containing CSV files
        dry_run: If True, only print planned actions without making changes
        verbose: If True, print detailed progress
        remove_timezone: If True, remove timezone labels from CSV files before renaming
        rename_time_column: If True, rename 'time' column to 'datetime'
    """
    csv_files = [f for f in os.listdir(directory) if f.lower().endswith('.csv')]
    print(f"Found {len(csv_files)} CSV files to process...")

    renamed_count = 0
    tz_removed_count = 0
    column_renamed_count = 0
    already_named_count = 0
    errors = []
    planned = []

    for csv_file in csv_files:
        # Skip files that are already properly named
        if is_properly_named(csv_file):
            already_named_count += 1
            if verbose:
                print(f"SKIP (already named): {csv_file}")
            continue
        
        csv_path = os.path.join(directory, csv_file)
        
        # Remove timezone from CSV content if requested
        if remove_timezone and not dry_run:
            success, error = process_csv_remove_timezone(csv_path, inplace=True, rename_time_column=rename_time_column)
            if success:
                tz_removed_count += 1
                if rename_time_column:
                    column_renamed_count += 1
                if verbose and tz_removed_count % 50 == 0:
                    print(f"Processed {tz_removed_count} files...")
            elif error:
                errors.append(f"Processing failed for {csv_file}: {error}")
        
        # Get dates for renaming
        first_date, last_date = get_date_from_csv(csv_path)
        if first_date and last_date:
            # normalize to filename: MM_DD_YYYY.csv
            formatted_date = first_date.replace('/', '_')
            new_filename = f"{formatted_date}.csv"
            new_path = os.path.join(directory, new_filename)
            planned.append((csv_file, new_filename))
            if dry_run:
                if verbose:
                    print(f"DRY RUN: {csv_file} -> {new_filename}")
                continue
            try:
                if os.path.exists(new_path) and os.path.abspath(new_path) != os.path.abspath(csv_path):
                    if verbose:
                        print(f"Warning: {new_filename} already exists, skipping {csv_file}")
                    errors.append(f"Target exists: {new_filename}")
                    continue
                if os.path.abspath(csv_path) != os.path.abspath(new_path):
                    os.rename(csv_path, new_path)
                    if verbose:
                        print(f"Renamed: {csv_file} -> {new_filename}")
                    renamed_count += 1
                else:
                    if verbose:
                        print(f"Already correctly named: {csv_file}")
            except Exception as e:
                error_msg = f"Error renaming {csv_file}: {e}"
                print(error_msg)
                errors.append(error_msg)
        else:
            error_msg = f"Could not extract dates from {csv_file}"
            if verbose:
                print(error_msg)
            errors.append(error_msg)

    print(f"\n{'='*60}")
    print(f"Processing complete! (dry_run={dry_run})")
    print(f"Files already properly named (skipped): {already_named_count}")
    if remove_timezone and not dry_run:
        print(f"Timezones removed from: {tz_removed_count} files")
    if rename_time_column and not dry_run:
        print(f"'time' column renamed to 'datetime' in: {column_renamed_count} files")
    print(f"Planned renames: {len(planned)}; Completed: {renamed_count}; Errors: {len(errors)}")
    if errors:
        print(f"\nShowing first 10 errors:")
        for e in errors[:10]:
            print(" -", e)
        if len(errors) > 10:
            print(f" ... and {len(errors) - 10} more errors")
    return {
        "planned": len(planned), 
        "renamed": renamed_count, 
        "already_named": already_named_count,
        "tz_removed": tz_removed_count,
        "column_renamed": column_renamed_count,
        "errors": errors
    }


if __name__ == "__main__":
    # Confirm directory exists
    if not os.path.exists(csv_directory):
        print(f"Error: Directory {csv_directory} does not exist!")
    else:
        print(f"Processing CSV files in: {csv_directory}")
        print("NOTE: Timestamps will be converted to UTC, timezone labels removed,")
        print("      and 'time' column will be renamed to 'datetime'.")
        print("      Files already named in MM_DD_YYYY.csv format will be skipped.\n")
        
        # Run with dry_run=True first to preview
        # _preview = rename_csv_files(csv_directory, dry_run=True, verbose=True, remove_timezone=True, rename_time_column=True)
        # print('\n=== Preview Summary ===')
        # print(f"Files to process: {_preview['planned']}")
        # print(f"Already named: {_preview['already_named']}")
        # print(f"Errors encountered: {len(_preview['errors'])}")
        
        # Actually perform the processing
        _result = rename_csv_files(csv_directory, dry_run=False, verbose=True, remove_timezone=True, rename_time_column=True)
```

```python
# Required imports
from pathlib import Path
import shutil
import pandas as pd

# Timezone offset mappings (hours from UTC) for NYISO
NYISO_TIMEZONE_OFFSETS = {
    'EDT': -4,  # Eastern Daylight Time
    'EST': -5,  # Eastern Standard Time
}


def process_nyiso_csvs(
    source_dir=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_csv\nyiso_all",
    dry_run=True,
    backup=True,
    preview_limit=3,
    verbose=True,
    convert_to_utc=True,
):
    """Normalize NYISO CSVs:

    - Look for 'Time Stamp' (case-insensitive) and optional 'Time Zone' column.
    - Example input: '01/02/2002 00:00:00' and 'EST' -> convert to UTC and normalize.
    - Convert to column `datetime` with format 'MM-DD-YYYY HH-MM-SS' and make it the first column.
    - If convert_to_utc=True, adjusts timestamps based on timezone (EDT/EST) to UTC.
    - Overwrite files unless dry_run=True. If backup=True a `.bak` copy is stored next to the CSV before overwrite.
    - Drops the original Time Stamp and Time Zone columns after conversion.
    """
    src = Path(source_dir)
    summary = {"files_found": 0, "processed": 0, "skipped": 0, "utc_converted": 0, "errors": []}

    if not src.exists():
        msg = f"Source not found: {src}"
        summary["errors"].append(msg)
        if verbose:
            print(msg)
        return summary

    csv_paths = sorted(src.glob("*.csv"))
    summary["files_found"] = len(csv_paths)
    if verbose:
        print(f"Found {len(csv_paths)} CSV files in {src}")
        if convert_to_utc:
            print("UTC conversion enabled - timestamps will be adjusted based on timezone")

    for p in csv_paths:
        try:
            df = pd.read_csv(p, dtype=str)
            if df.shape[0] == 0:
                summary['skipped'] += 1
                if verbose:
                    print(f"Skipping empty file: {p.name}")
                continue

            # find Time Stamp column (case-insensitive)
            cols_lower = [c.lower() for c in df.columns]
            ts_col = None
            for candidate in ('time stamp','timestamp','time'):
                if candidate in cols_lower:
                    ts_col = df.columns[cols_lower.index(candidate)]
                    break
            if ts_col is None:
                ts_col = df.columns[0]
                if verbose:
                    print(f"No 'Time Stamp' column in {p.name}; falling back to first column: {ts_col}")

            # find Time Zone column
            tz_col = None
            for c in df.columns:
                if str(c).strip().lower() in ('time zone', 'timezone', 'tz'):
                    tz_col = c
                    break

            raw_ts = df[ts_col].astype(str).fillna('').str.strip()
            
            # Extract timezone info if available
            timezone_series = None
            if tz_col and tz_col in df.columns:
                timezone_series = df[tz_col].astype(str).fillna('').str.strip()
            else:
                # Try to extract timezone from timestamp string itself
                timezone_series = raw_ts.str.extract(r'\s+([A-Z]{2,4})$', expand=False).fillna('')
            
            # Clean timestamps (remove timezone suffix if present)
            raw_clean = raw_ts.str.replace(r"\s+[A-Za-z]{1,5}$", "", regex=True)

            # Vectorized parsing for NYISO timestamps
            dt_series = pd.to_datetime(raw_clean, format='%m/%d/%Y %H:%M:%S', errors='coerce')
            nyiso_formats = [
                '%m/%d/%Y %H:%M',
                '%m/%d/%Y',
                '%Y-%m-%d %H:%M:%S',
                '%Y-%m-%d',
            ]
            for fmt in nyiso_formats:
                missing = dt_series.isna()
                if not missing.any():
                    break
                parsed_try = pd.to_datetime(raw_clean[missing], format=fmt, errors='coerce')
                dt_series.loc[missing] = parsed_try
            if dt_series.isna().any():
                remaining = dt_series.isna()
                dt_series.loc[remaining] = pd.to_datetime(raw_clean[remaining], errors='coerce')

            parsed_count = int(dt_series.notna().sum())
            if parsed_count == 0:
                msg = f"No time values parsed for {p.name}"
                summary['errors'].append(msg)
                if verbose:
                    print(msg)
                continue

            # Convert to UTC if requested
            if convert_to_utc:
                utc_converted_count = 0
                for idx in dt_series.index:
                    if pd.notna(dt_series[idx]):
                        tz_abbrev = timezone_series[idx] if timezone_series is not None else ''
                        if tz_abbrev in NYISO_TIMEZONE_OFFSETS:
                            offset_hours = NYISO_TIMEZONE_OFFSETS[tz_abbrev]
                            # Subtract the offset to get UTC (e.g., EST is UTC-5, so add 5 hours)
                            dt_series[idx] = dt_series[idx] - pd.Timedelta(hours=offset_hours)
                            utc_converted_count += 1
                
                if utc_converted_count > 0:
                    summary['utc_converted'] += 1
                    if verbose:
                        print(f"  {p.name}: Converted {utc_converted_count} timestamps to UTC")

            # Format datetime
            formatted = dt_series.apply(lambda dt: '' if pd.isna(dt) else dt.strftime('%m-%d-%Y %H-%M-%S'))

            df_new = df.copy()
            df_new['datetime'] = formatted

            to_drop = []
            if ts_col in df_new.columns:
                to_drop.append(ts_col)
            if tz_col and tz_col in df_new.columns:
                to_drop.append(tz_col)
            if to_drop:
                df_new = df_new.drop(columns=to_drop)

            # move datetime to first column
            cols = ['datetime'] + [c for c in df_new.columns if c != 'datetime']
            df_new = df_new[cols]

            if dry_run:
                if verbose:
                    print(f"DRY RUN - {p.name}: show first {preview_limit} rows of converted 'datetime' column:")
                    print(df_new[['datetime']].head(preview_limit))
                summary['processed'] += 1
                continue

            if backup:
                bak = p.with_suffix(p.suffix + '.bak')
                if not bak.exists():
                    shutil.copy2(p, bak)
                    if verbose:
                        print(f"Backup created: {bak.name}")

            df_new.to_csv(p, index=False)
            if verbose:
                print(f"Updated: {p.name} (wrote {len(df_new)} rows)")
            summary['processed'] += 1

        except Exception as e:
            msg = f"Error processing {p.name}: {e}"
            summary['errors'].append(msg)
            if verbose:
                print(msg)

    print(f"\n{'='*60}")
    print(f"Processing Summary:")
    print(f"  Files found: {summary['files_found']}")
    print(f"  Files processed: {summary['processed']}")
    if convert_to_utc:
        print(f"  Files with UTC conversion: {summary['utc_converted']}")
    print(f"  Files skipped: {summary['skipped']}")
    print(f"  Errors: {len(summary['errors'])}")
    
    return summary

# Preview with UTC conversion
_nyiso_preview = process_nyiso_csvs(dry_run=True, preview_limit=5, convert_to_utc=True)
```

## SORT BY YEAR 

```python
import os
import re
import shutil
from datetime import datetime
import json


def sort_csvs_by_year(
    source_dir=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\mesonet\mesonet_csv",
    dest_dir=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\mesonet\mesonet_csv",
    move=False,           # if True, move files; if False, copy files
    dry_run=False,         # safe default
    overwrite=False,
    verbose=True,
    patterns=None,
):
    """Sort CSV files into year subfolders under dest_dir.

    This function is robust to several filename date formats (e.g. MM_DD_YYYY, YYYYMMDD, YYYY_MM_DD).
    It returns a summary dict and (when dry_run=True) prints planned actions without touching the FS.
    Dates shown in the printed output are formatted as MM/DD/YYYY for readability.
    """
    if patterns is None:
        # regex patterns capturing year as a group named 'year'
        patterns = [
            re.compile(r"^(?P<month>\d{1,2})[_-](?P<day>\d{1,2})[_-](?P<year>\d{4})"),  # MM_DD_YYYY or M-D-YYYY
            re.compile(r"^(?P<year>\d{4})[_-]?(?P<month>\d{2})[_-]?(?P<day>\d{2})$"),   # YYYYMMDD or YYYY_MM_DD
            re.compile(r".*(?P<year>\d{4}).*"),  # fallback: any 4-digit year somewhere
        ]

    def _extract_mmddyyyy_from_name(name):
        """Try to extract month, day, year from filename stem and return 'MM/DD/YYYY' or None."""
        # Try same patterns used above
        for pat in patterns:
            m = pat.match(name)
            if m:
                gd = m.groupdict()
                if 'year' in gd and gd.get('month') and gd.get('day'):
                    try:
                        mm = int(gd['month'])
                        dd = int(gd['day'])
                        yy = int(gd['year'])
                        return f"{mm:02}/{dd:02}/{yy:04}"
                    except Exception:
                        pass
        # fallback: look for contiguous YYYYMMDD
        m2 = re.search(r"(?P<y>(19|20)\d{2})(?P<m>\d{2})(?P<d>\d{2})", name)
        if m2:
            try:
                mm = int(m2.group('m'))
                dd = int(m2.group('d'))
                yy = int(m2.group('y'))
                return f"{mm:02}/{dd:02}/{yy:04}"
            except Exception:
                pass
        # last-resort: find any 4-digit year and return Jan 01 of that year (not ideal)
        m3 = re.search(r"(19|20)\d{2}", name)
        if m3:
            try:
                yy = int(m3.group(0))
                return f"01/01/{yy:04}"
            except Exception:
                pass
        return None

    src = os.path.abspath(source_dir)
    dst = os.path.abspath(dest_dir)

    summary = {"files_found": 0, "planned": 0, "copied": 0, "moved": 0, "skipped": 0, "errors": []}

    if not os.path.exists(src):
        msg = f"Source directory not found: {src}"
        summary["errors"].append(msg)
        if verbose:
            print(msg)
        return summary

    # list CSVs in the top-level folder only
    csv_files = [f for f in os.listdir(src) if f.lower().endswith('.csv') and os.path.isfile(os.path.join(src, f))]
    summary["files_found"] = len(csv_files)

    if verbose:
        print(f"Found {len(csv_files)} CSV files in {src}")

    files_by_year = {}

    for fn in csv_files:
        name = os.path.splitext(fn)[0]
        year = None
        for pat in patterns:
            m = pat.match(name)
            if m:
                gd = m.groupdict()
                if 'year' in gd and gd['year']:
                    year = gd['year']
                    break
                # try to extract from named groups or positional
                groups = m.groups()
                for g in groups[::-1]:
                    if g and len(g) == 4 and g.isdigit():
                        year = g
                        break
                if year:
                    break
        if year is None:
            # try to find a 4-digit year anywhere in the filename
            m2 = re.search(r"(19|20)\d{2}", name)
            if m2:
                year = m2.group(0)
        if year is None:
            # could not parse year; record error but continue
            summary['errors'].append(f"Could not parse year from filename: {fn}")
            if verbose:
                print(f"Could not parse year from filename: {fn}")
            continue
        try:
            y_int = int(year)
        except Exception:
            summary['errors'].append(f"Parsed invalid year '{year}' from {fn}")
            continue
        files_by_year.setdefault(y_int, []).append(fn)

    if not files_by_year:
        if verbose:
            print("No files matched expected date patterns; nothing to move/copy.")
        return summary

    # show planned actions
    if dry_run and verbose:
        print("DRY RUN - planned actions:")
        for y in sorted(files_by_year):
            print(f" Year {y}: {len(files_by_year[y])} files -> {os.path.join(dst, str(y))}")
        # show a few examples with human-friendly dates (MM/DD/YYYY)
        for y in sorted(files_by_year)[:3]:
            for fn in files_by_year[y][:5]:
                srcp = os.path.join(src, fn)
                destp = os.path.join(dst, str(y), fn)
                pretty = _extract_mmddyyyy_from_name(os.path.splitext(fn)[0])
                date_note = f" (date: {pretty})" if pretty else ""
                print(f"  - {srcp} -> {destp}{date_note}")
        summary['planned'] = sum(len(v) for v in files_by_year.values())
        return summary

    # perform actions
    for y in sorted(files_by_year):
        year_dir = os.path.join(dst, str(y))
        os.makedirs(year_dir, exist_ok=True)
        for fn in files_by_year[y]:
            srcp = os.path.join(src, fn)
            destp = os.path.join(year_dir, fn)
            try:
                if os.path.exists(destp):
                    if overwrite:
                        os.remove(destp)
                    else:
                        summary['skipped'] += 1
                        if verbose:
                            print(f"Skipping existing: {destp}")
                        continue
                if move:
                    shutil.move(srcp, destp)
                    summary['moved'] += 1
                    action = 'Moved'
                else:
                    shutil.copy2(srcp, destp)
                    summary['copied'] += 1
                    action = 'Copied'
                if verbose:
                    print(f"{action}: {srcp} -> {destp}")
            except Exception as e:
                summary['errors'].append(f"Error handling {srcp}: {e}")
                if verbose:
                    print(f"Error handling {srcp}: {e}")

    if verbose:
        print(f"Done. Copied: {summary['copied']}, Moved: {summary['moved']}, Skipped: {summary['skipped']}, Errors: {len(summary['errors'])}")

    return summary


# Run a dry-run now to show what would happen
_preview = sort_csvs_by_year(dry_run=True, verbose=True)
print('\nPreview summary:\n', json.dumps(_preview, indent=2))

# RUN
# Execute the sorter to actually move files (destructive). Confirmed settings: move=True, dry_run=False, overwrite=False
res_move = sort_csvs_by_year(move=False, dry_run=False, overwrite=False, verbose=True)
print('\nMove result:\n', json.dumps(res_move, indent=2))
```

## MESONET ALL 

```python
import os
from pathlib import Path
import shutil

def gather_top_level_csvs(
    source_dir=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\mesonet\mesonet_csv",
    dest_dir=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\mesonet\mesonet_csv\mesonet_all",
    move: bool = False,
    dry_run: bool = False,
    overwrite: bool = True,
    verbose: bool = True
):
    """
    Move or copy CSV files that are directly inside `source_dir` (not inside any subfolders)
    into `dest_dir` as individual files (no aggregation).

    Parameters:
      - source_dir: folder containing nyiso CSVs (top-level files considered)
      - dest_dir: folder where top-level CSVs will be copied/moved
      - move: if True, move files; if False, copy files
      - dry_run: if True, only print planned actions and return summary (no fs changes)
      - overwrite: if True, overwrite existing files in dest_dir
      - verbose: if True, print progress

    Returns:
      summary dict: {files_found, files_transferred, skipped, errors}
    """
    src = Path(source_dir).expanduser().resolve()
    dst = Path(dest_dir).expanduser().resolve()
    summary = {"files_found": 0, "files_transferred": 0, "skipped": 0, "errors": []}

    if not src.exists():
        msg = f"Source directory not found: {src}"
        summary["errors"].append(msg)
        if verbose:
            print(msg)
        return summary

    # create destination (but don't create it when dry_run is True unless necessary to show path)
    if not dry_run:
        dst.mkdir(parents=True, exist_ok=True)

    # collect only top-level csv files (do not recurse)
    top_level = [p for p in src.iterdir() if p.is_file() and p.suffix.lower() == ".csv"]
    summary["files_found"] = len(top_level)
    if verbose:
        print(f"Found {len(top_level)} top-level CSV file(s) in {src}")

    if dry_run:
        if verbose:
            print("DRY RUN - actions shown below (no files will be moved/copied):")
            for p in top_level:
                print(f"  - {p.name} -> {dst / p.name}")
        return summary

    for p in top_level:
        dest_path = dst / p.name
        try:
            if dest_path.exists():
                if overwrite:
                    # remove existing before action
                    dest_path.unlink()
                else:
                    summary["skipped"] += 1
                    if verbose:
                        print(f"Skipping existing file: {dest_path.name}")
                    continue

            if move:
                shutil.move(str(p), str(dest_path))
                action = "Moved"
            else:
                shutil.copy2(str(p), str(dest_path))
                action = "Copied"

            summary["files_transferred"] += 1
            if verbose:
                print(f"{action}: {p.name} -> {dest_path}")
        except Exception as e:
            msg = f"Error handling {p.name}: {e}"
            summary["errors"].append(msg)
            if verbose:
                print(msg)

    if verbose:
        print(f"Done. {summary['files_transferred']} files {'moved' if move else 'copied'}; {summary['skipped']} skipped; {len(summary['errors'])} errors.")
    return summary

    # Execute the gather operation (copies top-level CSVs into mesonet_all).
    # Use the previous sorter result to decide whether to move files; default to copy.

gather_result = gather_top_level_csvs(move=False, dry_run=False, overwrite=True, verbose=True)
print("\nGather result:\n", gather_result)
```

### Convert to Parquets

```python
# Convert all CSVs under nyiso_csv -> nyiso_parquet (dry-run by default)
import os
from pathlib import Path
import pandas as pd


def convert_csvs_to_parquet(
    source_root=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\mesonet\mesonet_csv",
    dest_root=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\mesonet\mesonet_parquet",
    dry_run= False,
    overwrite=False,
    verbose=True,
    max_files=None,  # limit for testing
    preview_limit=50,  # number of file examples to show on dry-run
):
    """Recursively copy folder structure from source_root into dest_root and convert each .csv to .parquet.

    - dry_run=True will only print planned actions and return a summary.
    - overwrite controls whether existing .parquet files are replaced.
    - max_files can be set for testing to limit actual work.
    """
    src = Path(source_root).expanduser().resolve()
    dst = Path(dest_root).expanduser().resolve()
    summary = {"files_found": 0, "converted": 0, "skipped": 0, "errors": []}

    if not src.exists():
        msg = f"Source directory not found: {src}"
        summary["errors"].append(msg)
        if verbose:
            print(msg)
        return summary

    csv_paths = sorted(src.rglob("*.csv"))
    summary["files_found"] = len(csv_paths)

    if verbose:
        print(f"Found {len(csv_paths)} CSV file(s) under {src}")

    if len(csv_paths) == 0:
        return summary

    # show preview when dry-run
    if dry_run:
        print("DRY RUN - planned conversion (no files changed):")
        print(f"  - source root: {src}")
        print(f"  - dest root:   {dst}")
        if len(csv_paths) > 0:
            print(f"  - showing up to {preview_limit} files:")
            for p in csv_paths[:preview_limit]:
                rel = p.relative_to(src)
                out_path = (dst / rel).with_suffix('.parquet')
                print(f"    - {p}  ->  {out_path}")
            if len(csv_paths) > preview_limit:
                print(f"    ... and {len(csv_paths)-preview_limit} more files")
        return summary

    # ensure destination exists (we'll create subdirectories as needed)
    dst.mkdir(parents=True, exist_ok=True)

    processed = 0
    for p in csv_paths:
        if max_files is not None and processed >= max_files:
            break
        try:
            rel = p.relative_to(src)
            out_dir = (dst / rel).parent
            out_dir.mkdir(parents=True, exist_ok=True)
            out_file = (dst / rel).with_suffix('.parquet')

            if out_file.exists():
                if not overwrite:
                    summary["skipped"] += 1
                    if verbose:
                        print(f"Skipping existing: {out_file}")
                    processed += 1
                    continue
                else:
                    try:
                        out_file.unlink()
                    except Exception as e:
                        msg = f"Unable to remove existing output {out_file}: {e}"
                        summary["errors"].append(msg)
                        if verbose:
                            print(msg)
                        processed += 1
                        continue

            # read csv and write parquet
            try:
                df = pd.read_csv(p)
            except Exception as e:
                msg = f"Error reading CSV {p}: {e}"
                summary["errors"].append(msg)
                if verbose:
                    print(msg)
                processed += 1
                continue

            try:
                # Try default engine (pyarrow preferred)
                df.to_parquet(out_file, index=False)
            except Exception as e1:
                # attempt fallback to fastparquet if available
                try:
                    df.to_parquet(out_file, index=False, engine="fastparquet")
                except Exception as e2:
                    msg = f"Error writing parquet for {p}: {e1} | {e2}"
                    summary["errors"].append(msg)
                    if verbose:
                        print(msg)
                    processed += 1
                    continue

            summary["converted"] += 1
            if verbose:
                print(f"Converted: {p} -> {out_file}")

        except Exception as e:
            msg = f"Unexpected error processing {p}: {e}"
            summary["errors"].append(msg)
            if verbose:
                print(msg)
        processed += 1

    if verbose:
        print(f"Done. Converted: {summary['converted']}; Skipped: {summary['skipped']}; Errors: {len(summary['errors'])}")
    return summary


# Run a dry-run preview so you can inspect planned actions before converting.
_preview = convert_csvs_to_parquet(dry_run=False, preview_limit=50, verbose=True)
import json
print('\nPreview summary:\n', json.dumps(_preview, indent=2))



```

## NYISO AVERAGE

```python
#### CREATE NYISO MASTER

import os
from pathlib import Path
import pandas as pd
from datetime import datetime
import pytz

def combine_and_process_nyiso_master(
    source_dir=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_parquet\nyiso_all",
    dest_dir=r"c:\Users\Matt\Desktop\IntroDS\IntroDS_Project\1_LIB\nyiso\nyiso_parquet\nyiso_master",
    output_name="nyiso_master.parquet",
    dry_run=False,
    overwrite=False,
    verbose=True,
):
    """
    1. Combine all NYISO parquet files
    2. Average all numeric values by Time Stamp (excluding Time Stamp, Time Zone, Name)
    3. Rename 'Time Stamp' to 'datetime'
    4. Convert to UTC using Time Zone information
    5. Drop Time Zone and Name columns
    """
    src = Path(source_dir).expanduser().resolve()
    dst = Path(dest_dir).expanduser().resolve()
    summary = {"files_found": 0, "files_combined": 0, "errors": []}

    if not src.exists():
        msg = f"Source directory not found: {src}"
        summary["errors"].append(msg)
        if verbose:
            print(msg)
        return summary

    # Collect only top-level parquet files (do not recurse into subdirectories)
    parquet_files = sorted([p for p in src.glob("*.parquet") if p.is_file()])
    summary["files_found"] = len(parquet_files)

    if verbose:
        print(f"Found {len(parquet_files)} parquet file(s) in {src}")

    if len(parquet_files) == 0:
        print("No parquet files found in source directory")
        return summary

    out_path = dst / output_name

    if dry_run:
        if verbose:
            print("DRY RUN - planned actions:")
            print(f"  - destination: {out_path}")
            if out_path.exists():
                print(f"  - NOTE: output already exists: {out_path} (overwrite={overwrite})")
            print("  - files to be combined (showing up to 50):")
            for p in parquet_files[:50]:
                print(f"    - {p.name}")
            if len(parquet_files) > 50:
                print(f"    ... and {len(parquet_files)-50} more files")
        return summary

    # Ensure destination exists
    dst.mkdir(parents=True, exist_ok=True)

    # Check for existing output file
    if out_path.exists():
        if not overwrite:
            msg = f"Output file already exists: {out_path} (set overwrite=True to replace)"
            summary["errors"].append(msg)
            if verbose:
                print(msg)
            return summary
        else:
            try:
                out_path.unlink()
                if verbose:
                    print(f"Removed existing output: {out_path}")
            except Exception as e:
                msg = f"Unable to remove existing output: {e}"
                summary["errors"].append(msg)
                if verbose:
                    print(msg)
                return summary

    # STEP 1: Load all parquet files
    all_dfs = []
    combined_count = 0

    if verbose:
        print("\n" + "=" * 60)
        print("STEP 1: Loading parquet files...")
        print("=" * 60)

    for i, p in enumerate(parquet_files):
        try:
            df = pd.read_parquet(p)
            all_dfs.append(df)
            combined_count += len(df)
            if verbose and (i + 1) % 100 == 0:
                print(f"  [{i+1}/{len(parquet_files)}] Loaded {i+1} files...")
        except Exception as e:
            msg = f"Error loading {p.name}: {e}"
            summary["errors"].append(msg)
            if verbose:
                print(f"  ERROR: {msg}")

    if len(all_dfs) == 0:
        msg = "No data to combine"
        summary["errors"].append(msg)
        if verbose:
            print(msg)
        return summary

    # Combine all dataframes
    if verbose:
        print(f"\nCombining {len(all_dfs)} dataframes...")
    
    combined_df = pd.concat(all_dfs, ignore_index=True)
    
    if verbose:
        print(f"Combined shape: {combined_df.shape}")
        print(f"Columns: {list(combined_df.columns)}")

    # STEP 2: Average by Time Stamp
    if verbose:
        print("\n" + "=" * 60)
        print("STEP 2: Averaging by Time Stamp...")
        print("=" * 60)

    # Identify columns to exclude from averaging
    exclude_cols = {'Time Stamp', 'Time Zone', 'Name'}
    
    # Find which excluded columns actually exist
    existing_exclude = [col for col in exclude_cols if col in combined_df.columns]
    
    if verbose:
        print(f"Excluding from averaging: {existing_exclude}")

    # Identify numeric columns to average
    numeric_cols = []
    for col in combined_df.columns:
        if col not in exclude_cols:
            if pd.api.types.is_numeric_dtype(combined_df[col]):
                numeric_cols.append(col)

    if verbose:
        print(f"Averaging {len(numeric_cols)} numeric columns")
        if numeric_cols:
            print(f"Sample columns: {numeric_cols[:min(5, len(numeric_cols))]}")

    # Group by Time Stamp and average
    if 'Time Stamp' not in combined_df.columns:
        msg = "ERROR: 'Time Stamp' column not found!"
        summary["errors"].append(msg)
        if verbose:
            print(msg)
        return summary

    # Build aggregation dictionary
    agg_dict = {col: 'mean' for col in numeric_cols}
    
    # Keep first occurrence of Time Zone and Name if they exist
    if 'Time Zone' in combined_df.columns:
        agg_dict['Time Zone'] = 'first'
    if 'Name' in combined_df.columns:
        agg_dict['Name'] = 'first'

    averaged_df = combined_df.groupby('Time Stamp', as_index=False).agg(agg_dict)

    if verbose:
        print(f"After averaging shape: {averaged_df.shape}")
        print(f"Unique timestamps: {len(averaged_df)}")

    # STEP 3: Rename 'Time Stamp' to 'datetime'
    if verbose:
        print("\n" + "=" * 60)
        print("STEP 3: Renaming 'Time Stamp' to 'datetime'...")
        print("=" * 60)

    averaged_df = averaged_df.rename(columns={'Time Stamp': 'datetime'})
    
    if verbose:
        print("Column renamed successfully")

    # STEP 4: Convert to UTC using Time Zone
    if verbose:
        print("\n" + "=" * 60)
        print("STEP 4: Converting to UTC...")
        print("=" * 60)

    if 'Time Zone' in averaged_df.columns:
        # Parse datetime strings and convert to UTC
        def convert_to_utc(row):
            try:
                dt_str = str(row['datetime'])
                tz_str = str(row['Time Zone'])
                
                # Parse the datetime
                if pd.isna(dt_str) or pd.isna(tz_str):
                    return pd.NaT
                
                # Try to parse the datetime string
                dt = pd.to_datetime(dt_str)
                
                # Map common timezone abbreviations
                tz_map = {
                    'EST': 'US/Eastern',
                    'EDT': 'US/Eastern',
                    'CST': 'US/Central',
                    'CDT': 'US/Central',
                    'MST': 'US/Mountain',
                    'MDT': 'US/Mountain',
                    'PST': 'US/Pacific',
                    'PDT': 'US/Pacific',
                }
                
                tz_name = tz_map.get(tz_str, tz_str)
                
                # Localize to the timezone
                tz = pytz.timezone(tz_name)
                dt_localized = tz.localize(dt) if dt.tzinfo is None else dt
                
                # Convert to UTC
                dt_utc = dt_localized.astimezone(pytz.UTC)
                
                return dt_utc
            except Exception as e:
                if verbose:
                    print(f"  Warning: Could not convert {row['datetime']} ({row['Time Zone']}): {e}")
                return pd.NaT
        
        if verbose:
            print("Converting timestamps to UTC...")
        
        averaged_df['datetime'] = averaged_df.apply(convert_to_utc, axis=1)
        
        if verbose:
            print(f"Converted {averaged_df['datetime'].notna().sum()} timestamps to UTC")
    else:
        if verbose:
            print("No 'Time Zone' column found, ensuring datetime is proper type...")
        averaged_df['datetime'] = pd.to_datetime(averaged_df['datetime'])

    # STEP 5: Drop Time Zone and Name columns
    if verbose:
        print("\n" + "=" * 60)
        print("STEP 5: Dropping Time Zone and Name columns...")
        print("=" * 60)

    cols_to_drop = []
    if 'Time Zone' in averaged_df.columns:
        cols_to_drop.append('Time Zone')
    if 'Name' in averaged_df.columns:
        cols_to_drop.append('Name')

    if cols_to_drop:
        averaged_df = averaged_df.drop(columns=cols_to_drop)
        if verbose:
            print(f"Dropped columns: {cols_to_drop}")
    else:
        if verbose:
            print("No columns to drop")

    if verbose:
        print(f"Final shape: {averaged_df.shape}")
        print(f"Final columns: {list(averaged_df.columns)}")
        print(f"\nFirst few rows:")
        print(averaged_df.head())

    # STEP 6: Write to parquet
    if verbose:
        print("\n" + "=" * 60)
        print("STEP 6: Writing to parquet...")
        print("=" * 60)
        print(f"Output: {out_path}")

    averaged_df.to_parquet(out_path, index=False)

    summary["files_combined"] = len(all_dfs)
    if verbose:
        file_size = out_path.stat().st_size
        print(f"\nDone! Wrote {len(averaged_df):,} rows to {out_path}")
        print(f"File size: {file_size:,} bytes ({file_size / 1024 / 1024:.2f} MB)")
        print(f"Errors: {len(summary['errors'])}")
    
    return summary

# Run the processing
print("=" * 60)
print("NYISO MASTER PROCESSING")
print("=" * 60)
res = combine_and_process_nyiso_master(dry_run=False, overwrite= True, verbose=True)
print("\n" + "=" * 60)
import json
print("Result:\n", json.dumps(res, indent=2))
```

```python

# Combine all parquets from mesonet_master and nyiso_master into a single master parquet (minute-level)
# Notebook-friendly: no argparse, uses workspace paths directly (overridable by setting variables above)
from pathlib import Path
import pandas as pd
import sys

# Default directories (edit these if you want to run against other paths)
mesonet_dir = Path(r"c:\Users\Matt\Desktop\CS506\CS506_Project\1_LIB\mesonet\mesonet_parquet\mesonet_master")
nyiso_dir   = Path(r"c:\Users\Matt\Desktop\CS506\CS506_Project\1_LIB\nyiso\nyiso_parquet\nyiso_master")
out_dir     = Path(r"c:\Users\Matt\Desktop\CS506\CS506_Project\1_LIB\master")

print(f"mesonet_dir={mesonet_dir}")
print(f"nyiso_dir={nyiso_dir}")
print(f"out_dir={out_dir}")

def find_parquet_files(d: Path):
    if not d.exists():
        return []
    return sorted([p for p in d.rglob("*.parquet") if p.is_file()])

def read_and_concat_parquets(dirpath: Path):
    files = find_parquet_files(dirpath)
    if not files:
        print(f"Warning: no parquet files found in {dirpath}")
        return pd.DataFrame()

    parts = []
    for f in files:
        try:
            df = pd.read_parquet(f)
            # remember source file relative path
            try:
                df['_source_file'] = str(f.relative_to(dirpath))
            except Exception:
                df['_source_file'] = str(f)
            parts.append(df)
        except Exception as e:
            print(f"Error reading {f}: {e}", file=sys.stderr)
    if not parts:
        return pd.DataFrame()
    combined = pd.concat(parts, ignore_index=True, sort=False)
    return combined

def detect_datetime_column(df: pd.DataFrame):
    if df.empty:
        return None
    names = list(df.columns)
    # prioritized exact matches
    for cand in ("datetime","timestamp","time","date","ts","DateTime","DATE_TIME"):
        for c in names:
            if c.lower() == cand.lower():
                return c
    # fallback: first column containing date/time
    for c in names:
        if "date" in c.lower() or "time" in c.lower() or c.lower().endswith("_ts"):
            return c
    return None

# Read datasets
mesonet_df = read_and_concat_parquets(mesonet_dir)
nyiso_df   = read_and_concat_parquets(nyiso_dir)

print(f"Read {len(mesonet_df)} mesonet rows and {len(nyiso_df)} nyiso rows")

if mesonet_df.empty and nyiso_df.empty:
    print("No data to combine; aborting.")
else:
    # REMOVE 'station' COLUMN if it exists
    print("\nRemoving 'station' column if present...")
    for col in ['station', 'Station', 'STATION']:
        if col in mesonet_df.columns:
            mesonet_df = mesonet_df.drop(columns=[col])
            print(f"  Dropped '{col}' from mesonet data")
        if col in nyiso_df.columns:
            nyiso_df = nyiso_df.drop(columns=[col])
            print(f"  Dropped '{col}' from nyiso data")

    # detect dt columns
    m_dt = detect_datetime_column(mesonet_df)
    n_dt = detect_datetime_column(nyiso_df)
    print(f"\nDetected mesonet datetime column: {m_dt}")
    print(f"Detected nyiso datetime column: {n_dt}")

    # rename detected columns to a canonical 'datetime'
    if m_dt and m_dt != 'datetime':
        mesonet_df = mesonet_df.rename(columns={m_dt: 'datetime'})
    if n_dt and n_dt != 'datetime':
        nyiso_df = nyiso_df.rename(columns={n_dt: 'datetime'})

    # coerce to datetime where present
    if 'datetime' in mesonet_df.columns:
        mesonet_df['datetime'] = pd.to_datetime(mesonet_df['datetime'], errors='coerce')
    if 'datetime' in nyiso_df.columns:
        nyiso_df['datetime'] = pd.to_datetime(nyiso_df['datetime'], errors='coerce')

    # ensure datetime column exists in both (fill with NaT if missing)
    if 'datetime' not in mesonet_df.columns:
        mesonet_df['datetime'] = pd.NaT
    if 'datetime' not in nyiso_df.columns:
        nyiso_df['datetime'] = pd.NaT

    # NORMALIZE TIMEZONES - convert both to UTC timezone-aware
    print("\nNormalizing timezones to UTC...")
    if 'datetime' in mesonet_df.columns:
        if mesonet_df['datetime'].dt.tz is None:
            print("  Mesonet datetime is timezone-naive, converting to UTC")
            mesonet_df['datetime'] = mesonet_df['datetime'].dt.tz_localize('UTC')
        elif str(mesonet_df['datetime'].dt.tz) != 'UTC':
            print(f"  Mesonet datetime has timezone {mesonet_df['datetime'].dt.tz}, converting to UTC")
            mesonet_df['datetime'] = mesonet_df['datetime'].dt.tz_convert('UTC')
        else:
            print("  Mesonet datetime already in UTC")

    if 'datetime' in nyiso_df.columns:
        if nyiso_df['datetime'].dt.tz is None:
            print("  NYISO datetime is timezone-naive, converting to UTC")
            nyiso_df['datetime'] = nyiso_df['datetime'].dt.tz_localize('UTC')
        elif str(nyiso_df['datetime'].dt.tz) != 'UTC':
            print(f"  NYISO datetime has timezone {nyiso_df['datetime'].dt.tz}, converting to UTC")
            nyiso_df['datetime'] = nyiso_df['datetime'].dt.tz_convert('UTC')
        else:
            print("  NYISO datetime already in UTC")

    # ALIGN TO MINUTE for merge (use floor to avoid rounding up across minutes)
    print("\nAligning timestamps to minute for alignment (floor to 'min')...")
    if 'datetime' in mesonet_df.columns:
        mesonet_df['datetime'] = mesonet_df['datetime'].dt.floor('min')
        print(f"  Mesonet datetime range: {mesonet_df['datetime'].min()} to {mesonet_df['datetime'].max()}")
        print(f"  Mesonet unique minute timestamps: {mesonet_df['datetime'].nunique()}")

    if 'datetime' in nyiso_df.columns:
        nyiso_df['datetime'] = nyiso_df['datetime'].dt.floor('min')
        print(f"  NYISO datetime range: {nyiso_df['datetime'].min()} to {nyiso_df['datetime'].max()}")
        print(f"  NYISO unique minute timestamps: {nyiso_df['datetime'].nunique()}")

    # Remove duplicates after flooring (one row per minute)
    print("\nRemoving duplicate minute timestamps (ensures one row per minute)...")
    mesonet_df = mesonet_df.drop_duplicates(subset=['datetime'], keep='first')
    nyiso_df   = nyiso_df.drop_duplicates(subset=['datetime'], keep='first')
    print(f"  Mesonet rows after dedup: {len(mesonet_df)}")
    print(f"  NYISO rows after dedup: {len(nyiso_df)}")

    # reset indices
    mesonet_df = mesonet_df.reset_index(drop=True)
    nyiso_df   = nyiso_df.reset_index(drop=True)

    # perform INNER merge on datetime (ONLY keep matching minute timestamps in BOTH datasets)
    print("\nMerging datasets on datetime (INNER join - only minutes present in BOTH datasets)...")
    merged = pd.merge(mesonet_df, nyiso_df, on='datetime', how='inner', suffixes=('_mesonet','_nyiso'))
    print(f"Merged rows (inner join): {len(merged)}")

    # Verify one row per minute
    datetime_counts = merged['datetime'].value_counts()
    if (datetime_counts > 1).any():
        print(f"  WARNING: Found {(datetime_counts > 1).sum()} duplicate minute datetimes!")
    else:
        print(f"  ✓ Confirmed: One row per minute datetime")

    # try to sort by datetime
    if 'datetime' in merged.columns:
        try:
            merged = merged.sort_values('datetime')
            print(f"Sorted by datetime")
        except Exception as e:
            print(f"Could not sort by datetime: {e}")

    # prepare output
    out_dir.mkdir(parents=True, exist_ok=True)
    out_parquet = out_dir / 'master.parquet'

    print(f"\nWriting combined parquet to {out_parquet}")
    try:
        merged.to_parquet(out_parquet, index=False)
        print(f"Successfully wrote {len(merged)} rows to {out_parquet}")
        file_size = out_parquet.stat().st_size
        print(f"File size: {file_size:,} bytes ({file_size / 1024 / 1024:.2f} MB)")
    except Exception as e:
        print(f"Failed to write parquet: {e}")

    # show a quick preview
    print("\n" + "=" * 60)
    print("MERGE SUMMARY")
    print("=" * 60)
    print(f"Total rows: {len(merged)}")
    print(f"Total columns: {len(merged.columns)}")
    print(f"Datetime range: {merged['datetime'].min()} to {merged['datetime'].max()}")
    print(f"Unique minute datetimes: {merged['datetime'].nunique()}")
    print(f"Mesonet columns: {sum(1 for c in merged.columns if c.endswith('_mesonet'))}")
    print(f"NYISO columns: {sum(1 for c in merged.columns if c.endswith('_nyiso'))}")

    if len(merged) > 0:
        print("\nFirst 5 rows:")
        try:
            display(merged.head())
        except Exception:
            print(merged.head().to_string(index=False))

        print(f"\nAll columns ({len(merged.columns)}):")
        print(list(merged.columns))
    else:
        print("\nWARNING: No matching minute timestamps found between datasets!")
```

```python

import pandas as pd

# Ensure datetime column is proper dtype
df['datetime'] = pd.to_datetime(df['datetime'])

# Sort to ensure chronological order (recommended)
df = df.sort_values('datetime')

# Function to get first and last datetime where column is non-null
def col_start_end(df, time_col='datetime'):
    results = []
    for col in df.columns:
        if col == time_col:
            continue
        # Mask for non-null rows for this column
        mask = df[col].notna()
        if mask.any():
            start = df.loc[mask, time_col].iloc[0]
            end = df.loc[mask, time_col].iloc[-1]
            count_non_null = mask.sum()
        else:
            start = pd.NaT
            end = pd.NaT
            count_non_null = 0
        results.append({
            'column': col,
            'start_datetime': start,
            'end_datetime': end,
            'non_null_count': count_non_null
        })
    return pd.DataFrame(results).sort_values('column')

summary = col_start_end(df, time_col='datetime')

print(summary)
```
